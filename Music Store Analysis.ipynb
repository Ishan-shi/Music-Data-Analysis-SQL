```python
cus_inv= cus.merge(inv,on='customer_id')
```

```python
country_revenue= (
    cus_inv.groupby('country')['total'].sum().sort_values(ascending= False))

plt.figure(figsize=(12,6))

sns.barplot(x=country_revenue.values[:10],y=country_revenue.index[:10])
plt.title('Top Revenue Generating Countries')
plt.xlabel('Revenue')
plt.ylabel('Country')
plt.show()
```

```python
top_customers= (cus_inv.groupby(['first_name','last_name'])['total'].sum().sort_values(ascending=False))
top_customers.head(10)
```

```python
top10=top_customers.head(10)
plt.figure(figsize=(12,6))

sns.barplot(x=top10.values,y=[f'{idx[0]} {idx[1]}'
                                      for idx in top10.index])

plt.title('Top 10 Highest Value Customers')
plt.xlabel('Total Spending')
plt.ylabel('Customers')
plt.show()
```

```python
genre_sales=(il.merge(trk,on='track_id').merge(gen,on='genre_id'))
```

```python
gen_rev= genre_sales.groupby('name_y')['unit_price_y'].sum().sort_values(ascending= False)

plt.figure(figsize=(12,6))
sns.barplot(y=gen_rev.values[:10],x=gen_rev.index[:10])
plt.title('Top Revenue Generationg Genres')
plt.xlabel('Revenue')
plt.ylabel('Genre')
plt.show()
```

```python
artist_sales= (il.merge(trk,on='track_id').merge(alb,on='album_id').merge(art,on='artist_id'))
```

```python
artist_sales = artist_sales.rename(columns= {'unit_price_x':'invoice_price',
                                             'name_y':'artist_name'})
```

```python
top_artists= (artist_sales.groupby('artist_name')['invoice_price'].sum().sort_values(ascending= False))

plt.figure(figsize=(12,6))
sns.barplot(y=top_artists.values[:10],x=top_artists.index[:10])
plt.title("Top Revenue Generating Artists")
plt.xlabel('Artists')
plt.ylabel('Revenue')
plt.show()
```

```python
inv['invoice_date']= pd.to_datetime(inv['invoice_date'])

inv['Year']= inv['invoice_date'].dt.year

yearly_rev=(inv.groupby('Year')['total'].sum()
           )
```

```python
plt.figure(figsize=(12,6))

sns.lineplot(x=yearly_rev.index,y=yearly_rev.values,marker=0)
plt.title('Revenue Trend Over Time')
plt.xlabel('Year')
plt.ylabel('Revenue')
plt.show()
```

```python
inv['Month']= inv['invoice_date'].dt.month

mon_rev= (inv.groupby('Month')['total'].sum())
```

```python
plt.figure(figsize=(12,6))

sns.barplot(x=mon_rev.index,y=mon_rev.values)
plt.title('Revenue Trend Over Month')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.show()
```
